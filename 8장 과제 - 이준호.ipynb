## LAB 8-1: 오늘의 날짜와 현재 시간 출력하기
datetime 모듈을 사용하여 오늘의 날짜와 현재 시간을 오전/오후 형식으로 출력한다.


```python
import datetime as dt

now = dt.datetime.now()
today = dt.date.today()

if now.hour < 12:
    am_pm = '오전'
    hour = now.hour
else:
    am_pm = '오후'
    hour = now.hour - 12

if hour == 0:
    hour = 12

print(f'오늘의 날짜 : {today.year}년 {today.month}월 {today.day}일')
print(f'현재시간 : {am_pm} {hour}시 {now.minute}분 {now.second}초')
```

## LAB 8-2: 남은 날짜 계산하기
오늘 날짜를 기준으로 2027년 크리스마스, 2036년 새해, 다음 생일까지 남은 날짜와 시간을 계산한다.


```python
import datetime as dt

today = dt.date.today()
print(f'오늘은 {today.year}년 {today.month}월 {today.day}일입니다.')

xMas = dt.datetime(2027, 12, 25)
gap = xMas - dt.datetime.now()
print(f'2027년 크리스마스 까지는 {gap.days}일 {gap.seconds // 3600}시간 남았습니다.')

new_year = dt.datetime(2036, 1, 1)
gap = new_year - dt.datetime.now()
print(f'2036년 새해 까지는 {gap.days}일 {gap.seconds // 3600}시간 남았습니다.')

# 자신의 생일에 맞게 월과 일을 수정한다.
birthday = dt.datetime(today.year, 8, 29)
if birthday < dt.datetime.now():
    birthday = dt.datetime(today.year + 1, 8, 29)
gap = birthday - dt.datetime.now()
print(f'{birthday.year}년 생일까지는 {gap.days}일 {gap.seconds // 3600}시간 남았습니다.')
```

## LAB 8-3: 날짜 계산하기
PPT의 timedelta 예제처럼 날짜에 일정한 날짜 수를 더하여 1,000일 후 날짜와 100일 기념일을 계산한다.


```python
import datetime as dt

today = dt.datetime.now()
thousand = dt.timedelta(days=1000)
plus1000day = today + thousand
print('오늘 =', today)
print('1,000일 후 =', plus1000day)

start_text = input('처음으로 사귄 연도와 월, 일을 입력하시오 : ')
year, month, day = map(int, start_text.split())
start_day = dt.datetime(year, month, day)
hundred = dt.timedelta(days=99)
anniversary = start_day + hundred
print(f'100일 기념일은 : {anniversary.year}년 {anniversary.month}월 {anniversary.day}일입니다.')
```

## LAB 8-4: math 모듈을 이용한 계산
math 모듈과 for-in range()를 사용하여 4의 거듭제곱, 라디안 변환, sin 값을 출력한다.


```python
import math as m

for n in range(2, 11):
    print('4**{} = {:.1f}'.format(n, m.pow(4, n)))

print()
for degree in range(0, 181, 10):
    radian = m.radians(degree)
    print('{:3d} degree = {:.3f} radian'.format(degree, radian))

print()
for degree in range(0, 181, 10):
    print('sin({:3d}) = {:.2f}'.format(degree, m.sin(m.radians(degree))))
```

## LAB 8-5: 랜덤 번호 생성기
random 모듈의 randrange()와 sample()을 사용하여 임의의 정수를 출력한다.


```python
import random as rd

num_list = []
for _ in range(3):
    num_list.append(rd.randrange(0, 101, 5))
print('0에서 100 이하의 정수 중에서 5의 배수', num_list)

num_list = list(range(1, 11))
num_list = rd.sample(num_list, 3)
print('1에서 10 사이의 임의의 정수 :', num_list)
```

## LAB 8-6: 로또 번호 얻기
PPT의 로또 번호 만들기 코드처럼 list(range(1, 46)), rd.sample(), sort()를 사용한다.


```python
import random as rd

print('이번 주의 추천 로또 번호')
for _ in range(5):
    lotto_list = list(range(1, 46))       # 1부터 45까지 생성
    lotto_list = rd.sample(lotto_list, 6) # 임의의 값 6개 추출
    lotto_list.sort()                     # 선택된 번호를 정렬
    print(lotto_list)

lotto_list = list(range(1, 46))
winning = rd.sample(lotto_list, 6)
winning.sort()
bonus_list = [num for num in lotto_list if num not in winning]
bonus = rd.sample(bonus_list, 1)
print()
print('이번 주의 당첨 로또 번호 + 보너스')
print(winning, '+', bonus)
```

## 연습문제 8.1: 연애 시작일과 기념일 계산
2019년 2월 14일을 시작일로 두고 오늘까지 경과한 날짜와 100일, 200일, 500일, 1,000일 기념일을 계산한다.


```python
import datetime as dt

start_day = dt.date(2019, 2, 14)
today = dt.date.today()
gap = today - start_day

print(f'춘향이와 몽룡이의 연애 시작일 : {start_day.year}년 {start_day.month}월 {start_day.day}일')
print(f'연애 시작일로부터 경과한 날짜 : {gap.days}일')

for day in [100, 200, 500, 1000]:
    anniversary = start_day + dt.timedelta(days=day)
    print(f'{day:,}일 기념일 : {anniversary.year}년 {anniversary.month}월 {anniversary.day}일')
```

## 연습문제 8.2: 1부터 1,000,000까지의 합과 실행 시간
PPT의 elapsed_time.py 코드처럼 time.time()으로 시작 시간과 종료 시간을 기록하여 실행 시간을 계산한다.


```python
import time

def sum1to1000000():
    result = 0
    for i in range(1, 1000001):
        result = result + i
    return result

start_time = time.time()              # 시작 시간을 기록
sum1to1000000()
end_time = time.time()                # 종료 시간을 기록
gap = end_time - start_time
print('1에서 1,000,000까지의 합을 구하는 시간 : {:7.4f}초'.format(gap))

start_time = time.time()
for _ in range(100):
    sum1to1000000()
end_time = time.time()
gap = end_time - start_time
print('1,000,000까지의 합을 100번 반복해서 구하는 시간 : {:7.4f}초'.format(gap))
```

## 연습문제 8.3: 여러 함수의 100회 반복 실행 시간 측정
PPT의 시간 측정 코드 구조를 이용하여 여러 함수를 100번 반복 실행하는 데 걸리는 시간을 출력한다.


```python
import time
import math as m
import contextlib
import io

def hello_python():
    print('Hello Python!')

def factorial_10000():
    result = 1
    for i in range(1, 10001):
        result = result * i
    return result

def odd_cube_sum():
    result = 0
    for i in range(1, 10001, 2):
        result = result + i ** 3
    return result

def sin_sum():
    result = 0
    for degree in range(1, 361):
        result = result + m.sin(m.radians(degree))
    return result

def check_time(func):
    start_time = time.time()
    for _ in range(100):
        func()
    end_time = time.time()
    return end_time - start_time

with contextlib.redirect_stdout(io.StringIO()):
    gap = check_time(hello_python)
print('Hello Python! 출력을 100번 반복하는 시간 : {:7.4f}초'.format(gap))
print('10000!을 100번 반복해서 구하는 시간 : {:7.4f}초'.format(check_time(factorial_10000)))
print('1에서 10000까지 홀수의 세제곱 더하기를 100번 반복하는 시간 : {:7.4f}초'.format(check_time(odd_cube_sum)))
print('1에서 360도까지 sin 값의 합을 100번 반복하는 시간 : {:7.4f}초'.format(check_time(sin_sum)))
```

## 연습문제 8.4: 의사난수 함수 만들기
